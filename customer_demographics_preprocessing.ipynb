```python
#Importing Library

import pandas as pd
```

```python
#Load Datasets

df = pd.read_csv("customer_demographics_contaminated.csv")
```

```python
df
```

```python
#Dropping duplicate rows

df = df.drop_duplicates()
df
```

```python
df.describe()
```

```python
# checking for null 

df.isnull().sum()
```

```python
#Dropping rows with null values in Income Level column

df_clean = df.dropna(subset=["IncomeLevel"])
df_clean
```

```python
df_clean.isnull().sum()
```

```python
df_clean['Age'].value_counts()
```

```python
# Convert the Age column to numeric
df_clean['Age'] = pd.to_numeric(df_clean['Age'], errors='coerce')

# Calculate the iqr
q1 = df_clean['Age'].quantile(0.25)
q3 = df_clean['Age'].quantile(0.75)
iqr = q3 - q1
iqr
```

```python
# Calculating the lower limit and the Upper limit

lower_limit = q1 - iqr
upper_limit = q3 + iqr
lower_limit, upper_limit
```

```python
# Calculate the mean age
mean_age = df_clean['Age'].mean()

# Replaces NaN with mean age
df_clean['Age'] = df_clean['Age'].fillna(mean_age)

mean_age
```

```python
df_clean['Age'].value_counts()
```

```python
df_clean.isnull().sum()
```

```python
#Removes outliers in age that are below lower limit and above upper limit

df_clean = df_clean[(df_clean['Age'] >= lower_limit) & (df_clean['Age'] <= upper_limit)]
df_clean['Age'].value_counts()
```

```python
df_clean
```

```python
df_clean['Age'] = df_clean['Age'].round()
df_clean
```

```python
# Export the DataFrame to a CSV file
df_clean.to_csv('customer_demographics_cleaned.csv', index=False)
```
