```python
#Removing 2 transaction with a specific transaction ID. Amount is NaN for both transation and have different dates, Better to remove rows than to imputate mean

df = df[df["TransactionID"] != "62186b07-9cda-44c5-bf76-bdb0f97aaf39"]
```

```python
#Reloaded from the update above

duplicates_transaction = df[df["TransactionID"].duplicated(keep=False)].sort_values("TransactionID")

duplicates_transaction
```

```python
#Setting a standard date format of TransactionDate by making the datatype datetime

df["TransactionDate"] = pd.to_datetime(df["TransactionDate"], format="mixed", dayfirst=True, errors="coerce")
```

```python
#Showing 20 random rows to double check the TransactionDate Format

df.sample(20)
```

```python
#Checking for the unique values in "Amount" 
##Only the -100, free, and 0, are being analyzed at the moment

df["Amount"].value_counts()
```

```python
#Replacing the "Free" to 0 to make the values in "Amount" all numeric

df["Amount"] = df["Amount"].replace("Free", 0)
```

```python
#Converting the dtype of Amount column to numeric

df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")
```

```python
#Checking new info for the data type

df.info()
```

```python
#Checking the statistical values of numerical columns with describe

df.describe() 
```

```python
#Getting the mean and median with the amount using product category

df.groupby("ProductCategory")["Amount"].agg(["mean", "median" , "min", "max"])

#Analysis: All product category have the same min and almost similar max, the amount of difference in mean and median are also little
#Recommend: Still use the median for all and imputate missing values on the median of their respective product category
```

```python
#Checking if there is a row that has null value in both amount and product category

df[df["Amount"].isna() & df["ProductCategory"].isna()]
```

```python
#Checking the sum of rows that have null amount and null in product category 

(df["Amount"].isna() & df["ProductCategory"].isna()).sum()

#Analysis: transactions are shown with null amount and null product categories
#Recommend: Best to remove this rows since this is specifically a transaction csv/
```

```python
#Checking shape before dropping rows

df.shape
```

```python
#Dropping rows that have null values in both amount and productcategory

df = df[~(df["Amount"].isna() & df["ProductCategory"].isna())]

df.shape

#Analysis: 3013 - 26 rows = 2987, only 26 rows were removed. 
```

```python
#Checking if there are still rows with both amount and productcategory null

df[df["Amount"].isna() & df["ProductCategory"].isna()]

#Analysis: 0 rows were found
```

```python
#Checking the sum of null values in Amount

df["Amount"].isna().sum()
```

```python
#Creating a median variable for each Product Category

automotive_median = df[df["ProductCategory"] == "Automotive"]["Amount"].median()
clothing_median = df[df["ProductCategory"] == "Clothing"]["Amount"].median()
electronics_median = df[df["ProductCategory"] == "Electronics"]["Amount"].median()
health_beauty_median = df[df["ProductCategory"] == "Health & Beauty"]["Amount"].median()
home_garden_median = df[df["ProductCategory"] == "Home & Garden"]["Amount"].median()


print("Median of Product Categories")
print("The median for Automotive is: ", automotive_median)
print("The median for Clothing is: ", clothing_median)
print("The median for Electronics is: ", electronics_median)
print("The median for Health & Beauty is: ", health_beauty_median)
print("The median for Home & Garden is: ", home_garden_median)
```

```python
#Imputing null values of Amount with the mean of their corresponding Product Category

df.loc[df["ProductCategory"] == "Automotive", "Amount"] = df.loc[df["ProductCategory"] == "Automotive", "Amount"].fillna(automotive_median)
df.loc[df["ProductCategory"] == "Clothing", "Amount"] = df.loc[df["ProductCategory"] == "Clothing", "Amount"].fillna(clothing_median)
df.loc[df["ProductCategory"] == "Electronics", "Amount"] = df.loc[df["ProductCategory"] == "Electronics", "Amount"].fillna(electronics_median)
df.loc[df["ProductCategory"] == "Health & Beauty", "Amount"] = df.loc[df["ProductCategory"] == "Health & Beauty", "Amount"].fillna(health_beauty_median)
df.loc[df["ProductCategory"] == "Home & Garden", "Amount"] = df.loc[df["ProductCategory"] == "Home & Garden", "Amount"].fillna(home_garden_median)

#checking the sum of null values after imputing
df["Amount"].isna().sum()
```

```python
df["ProductCategory"].isna().sum()
```

```python
#Checking corelation with ProductCategory and PaymentMethod

df.groupby("ProductCategory")["PaymentMethod"].value_counts().sort_values

#Analysis, there is not hard evidence of the connection of a payment method to a specific product category
#Analysis 2: Since there are 256 row of missing values, imputing the mode would push 1 product category above others
#Recommend: Fill all the null values with new category "Unspecified" to avoid skewing the data
```

```python
#Filling null values in ProductCategory as Unspecified

df["ProductCategory"] = df["ProductCategory"].fillna("Unspecified")
```
