```python
# Refresh duplicated CustomerID records
duplicates = df[df["CustomerID"].duplicated(keep=False)]

# Refresh CustomerID conflicts
conflicts_CustomerID = duplicates.groupby("CustomerID").nunique()

conflicts_CustomerID = conflicts_CustomerID[(conflicts_CustomerID != 1).any(axis=1)]
```

```python
# Creating a variable for signup conflicts
signup_conflicts = conflicts_CustomerID[conflicts_CustomerID["SignupDate"] > 1]
```

```python
# Print duplicates where SignupDate does not match

print(duplicates[duplicates["CustomerID"].isin(signup_conflicts.index)][["CustomerID", "SignupDate"]].sort_values("CustomerID"))
```

```python
#Sorting the values of signupDate to make the earlier date on top

df = df.sort_values("SignupDate")

#Get CustomerIDs with conflicting SignupDates
signup_ids = signup_conflicts.index

#Combining the non-conflicting records with the corrected SignupDate records
df = pd.concat([
    df[~df["CustomerID"].isin(signup_ids)],
    df[df["CustomerID"].isin(signup_ids)].drop_duplicates(
        subset="CustomerID", keep="last"
    )
], ignore_index=True)

# Reason: Opted to keep the last since this is the latest data that the person made.
```
