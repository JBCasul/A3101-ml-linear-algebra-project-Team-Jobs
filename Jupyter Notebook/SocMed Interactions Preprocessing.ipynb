```python
#Dropping global duplicates

df = df.drop_duplicates()
```

```python
#Standardizing InteractionDate format

df["InteractionDate"] = pd.to_datetime(df["InteractionDate"], format="mixed", dayfirst=True, errors="coerce")
```

```python
#Checking for possible null values in InteractionDate after standardization

df["InteractionDate"].isna().sum()
```

```python
#Sorting dataset into InteractionDate and restarting the row number

df = df.sort_values("InteractionDate", ascending=True).reset_index(drop=True)

df
```

```python
#Checking to see if there are repeated Interaction ID

df["InteractionID"].duplicated().sum()
```

```python
#Creating a duplicates variable for the column InteractionID

duplicates = df[df["InteractionID"].duplicated(keep=False)]

duplicates
```

```python
#Creating a variable to view difference in InteractionID per column

conflicts_InteractionID = duplicates.groupby("InteractionID").nunique()

conflicts_InteractionID
```

```python
conflicts_InteractionID = conflicts_InteractionID[(conflicts_InteractionID != 1).any(axis=1)]

conflicts_InteractionID
```

```python
#Checking Unull values for Platform

df["Platform"].isna().sum()

#Recommend: Fill the na Platform to "Unspecified" and not mode to not skew the Platform distribution
```

```python
df.duplicated().sum()
```

```python
#Filling the null values in Platform with "Unspecified" 

df["Platform"] = df["Platform"].fillna("Unspecified")
```

```python
#Checking new Value counts of Platform

df["Platform"].value_counts()
```

```python
#Creating a variable for date conflicts

date_conflicts = conflicts_InteractionID[conflicts_InteractionID["InteractionDate"] > 1].index
```

```python
#Finding the later date for each date conflict

later_date_rows = df[df["InteractionID"].isin(date_conflicts)].groupby("InteractionID")["InteractionDate"].idxmax()
```

```python
#Dropping only the later date rows

df = df.drop(index=later_date_rows).sort_values("InteractionDate").reset_index(drop=True)
```

```python
#Refresh conflicts_InteractionID

duplicates = df[df["InteractionID"].duplicated(keep=False)]

conflicts_InteractionID = duplicates.groupby("InteractionID").nunique()

conflicts_InteractionID = conflicts_InteractionID[(conflicts_InteractionID != 1).any(axis=1)]

conflicts_InteractionID
```

```python
#Checking actual values of Sentiment from conflicts

df[df["InteractionID"].isin(conflicts_InteractionID.index)].sort_values("InteractionID")

#Analysis: Unable to create a justifiable way to choose which sentiment to retain and to drop, Flagging these rows for conflicting values
#Action: Will retain the data and treat this as two separate interaction from a customer despite having the same InteractionID
```

```python
#Checking for null values in Sentiment

df["Sentiment"].isna().sum()

#Analysis: 10% of data will be imputed, it will skew the EDA. 
#Recommend: Instead of imputing mode or dropping rows, fill null values with Unknown
```

```python
#Filling the null values with "Unknown"

df["Sentiment"] = df["Sentiment"].fillna("Unknown")
```
