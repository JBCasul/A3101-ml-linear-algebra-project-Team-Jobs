```python
#Import Libraries

import pandas as pd
```

```python
#Load Datasets

df = pd.read_csv("customer_transactions_contaminated.csv")
```

```python
df
```

```python
#Dropping duplicate rows

df = df.drop_duplicates()
df
```

```python
# checking for null 

df.isnull().sum()
```

```python
#Dropping rows with null values in Product Category column

df_clean = df.dropna(subset=["ProductCategory"])
df_clean
```

```python
df_clean.isnull().sum()
```

```python
df_clean = df_clean[df_clean['Amount'] > 0]
df_clean
```

```python
df_clean.isnull().sum()
```

```python
df['TransactionDate'] = pd.to_datetime(df['TransactionDate'], format='mixed', errors='coerce')

df['TransactionDate'] = df['TransactionDate'].dt.strftime('%Y/%m/%d')
```

```python
df_clean.dtypes
```

```python
# Standardize separators to slashes
s = df['TransactionDate'].astype(str).str.replace('-', '/')

# Create an empty series for the results
parsed_dates = pd.Series(pd.NaT, index=df.index)

# Group 1: Rows where the year is first (starts with 4 digits, e.g., 2026/...)
is_year_first = s.str.match(r'^\d{4}')
parsed_dates[is_year_first] = pd.to_datetime(s[is_year_first], errors='coerce')

# Group 2: Rows where the day is first (e.g., 31/12/2026)
is_day_first = ~is_year_first & s.notna()
parsed_dates[is_day_first] = pd.to_datetime(s[is_day_first], dayfirst=True, errors='coerce')

# Convert back to standard YYYY/MM/DD format string
df_clean['TransactionDate'] = parsed_dates.dt.strftime('%Y/%m/%d')
```
